````python
# Create another random tensor
tensor2=torch.rand(1, 7)
#transpose the tensor
tensor2=tensor2.T

# Perform matrix multiplication
torch.matmul(tensor1, tensor2)
````


### 6. Create two random tensors of shape `(2, 3)` and send them both to the GPU (you'll need access to a GPU for this). Set `torch.manual_seed(1234)` when creating the tensors (this doesn't have to be the GPU random seed). The output should be something like:

```
Device: cuda
(tensor([[0.0290, 0.4019, 0.2598],
         [0.3666, 0.0583, 0.7006]], device='cuda:0'),
 tensor([[0.0518, 0.4681, 0.6738],
         [0.3315, 0.7837, 0.5631]], device='cuda:0'))
```

````python
# Set random seed
torch.manual_seed(1234)

# Check for access to GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Device: {device}")

# Create two random tensors on GPU
tensor_A = torch.rand(size=(2, 3)).to(device)
tensor_B = torch.rand(size=(2, 3)).to(device)
tensor_A, tensor_B
````


### 7. Perform a matrix multiplication on the tensors you created in 6 (again, you may have to adjust the shapes of one of the tensors).

The output should look like:

````python
# Perform matmul on tensor_A and tensor_B
torch.matmul(tensor_A, tensor_B.T)
````

### 8. Find the maximum and minimum values of the output of 7.

````python
# Find max
torch.max(torch.matmul(tensor_A, tensor_B.T))

# Find min
torch.min(torch.matmul(tensor_A, tensor_B.T))
````

### 9. Find the maximum and minimum index values of the output of 7.

````python
# Find arg max
torch.argmax(torch.matmul(tensor_A, tensor_B.T))


# Find arg min
torch.argmin(torch.matmul(tensor_A, tensor_B.T))
````


### 10. Make a random tensor with shape `(1, 1, 1, 10)` and then create a new tensor with all the `1` dimensions removed to be left with a tensor of shape `(10)`. Set the seed to `7` when you create it and print out the first tensor and it's shape as well as the second tensor and it's shape.

The output should look like:

```
tensor([[[[0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297,
           0.3653, 0.8513]]]]) torch.Size([1, 1, 1, 10])
```

````python
# Set seed
torch.manual_seed(7)

# Create random tensor
tensor_A=torch.rand(1, 1, 1, 10)

# Remove single dimensions
tensor_A=tensor_A.squeeze()

# Print out tensors and their shapes
tensor_A, tensor_A.shape


# Remove single dimensions
torch.squeeze(tensor_A)


# Print out tensors and their shapes
tensor_A, tensor_A.shape
````
